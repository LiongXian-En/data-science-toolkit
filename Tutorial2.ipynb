```python
#Part 1-Google Colab and Python
#1. Create a new notebook and code the following

#1. Create a numpy array of size 2 x 2
import numpy as np

array_2x2 = np.array([
    [1, 2],
    [3, 4]
])
print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
#2. Create a random numpy array of size 3 x 4

random3x4 = np.random.rand(3, 4)

print("Random 3x4 array")
print(random3x4)
```

```python
#3. Create a numpy array of ones and a numpy array of zeroes
ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Ones array:")
print(ones_array)

print("\nZeros array:")
print(zeros_array)
```

```python
#4. Create a random numpy array of size 6x3 and transpose it

random6x3 = np.random.rand(6,3)
transposed_array = np.transpose(random6x3)

print("Original 6x3 Array\n",random6x3)

print("\nTransposed 6X3 Array (3x6 Array)\n",transposed_array)
```

```python
#5. Create two random numpy array of size n x p and p x m and then compute the dot
n,p,m = 3,4,2

A = np.random.rand(n,p)
B = np.random.rand(p,m)
dot_product = np.dot(A,B)

print("Array A:\n", A)
print("\nArray B:\n", B)
print("\nDot Product Result:\n", dot_product)
```
